```python
import os
import re


# Replace the 'parent_directory' with your directory path
parent_directory = "./mes-adresses"

target_string ="{% hint style=\"success\" %}"#"alt=\"\">"#"&#x20;"#"{% endhint %}"#"../.gitbook/assets/" #"<br>"##"alt=\"\">" #
replace_string =":::success"#"alt=\"\"/>"#":::"#"/img/mes-adresses/" #"<br />"#""#":::tip" #"/img" #":::warning"   #":::" # "alt=\"\"/>"          #

regex_search_term = " width=\"\d{3}\""
regex_replacement = "/"



for root, dirs, files in os.walk(parent_directory):
    for file in files:
        if file.endswith(".md"):
            print(os.path.join(root, file))
            with open(os.path.join(root, file), 'r') as f:
                lines = f.readlines()
            
            with open(os.path.join(root, file), 'w') as file:
                for line in lines:
                    newline = line.replace(target_string, replace_string)
                    #newline = re.sub(regex_search_term, regex_replacement, line)
                    file.write(newline)
```
