```python
import os
import re


# Replace the 'parent_directory' with your directory path
parent_directory = "./bonnes-pratiques"

target_string = "<br>" #"&#x20;"#"{% hint style=\"info\" %}"#"..//img" #"{% hint style=\"warning\" %}"  #"{% endhint %}"  #"alt=\"\">" #"../.gitbook/assets/"
replace_string = "<br />"#""#":::tip" #"/img" #":::warning"   #":::" # "alt=\"\"/>"          #"/img/"

for root, dirs, files in os.walk(parent_directory):
    for file in files:
        if file.endswith(".md"):
            print(os.path.join(root, file))
            with open(os.path.join(root, file), 'r') as f:
                lines = f.readlines()
            
            with open(os.path.join(root, file), 'w') as file:
                for line in lines:
                    newline = line.replace(target_string, replace_string)
                    file.write(newline)
```
